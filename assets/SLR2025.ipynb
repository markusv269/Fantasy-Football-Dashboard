## Ligenerstellung

### Gruppen erstellen

```python
# DataFrames vorbereiten
wunsch_dict = {}
commish_df = pd.DataFrame()
player_df = pd.DataFrame()

# Iteration über alle Zeilen des DataFrames
for index, row in df_slrleagues.iterrows():
    spieler = row["index"]

    # Wunsch-Mitspieler extrahieren
    mitspieler_liste = row["Mitspieler"]
    if pd.notna(mitspieler_liste):
        for mitspieler in mitspieler_liste.split(","):
            mitspieler = mitspieler.lower().strip()
            if mitspieler:  # nur nicht-leere Strings
                wunsch_dict.setdefault(spieler, []).append(mitspieler)

    # Aufteilen in Commishs und normale Spieler
    if row.get("Commish", False):
        commish_df = pd.concat([commish_df, row.to_frame().T], ignore_index=True)
    else:
        player_df = pd.concat([player_df, row.to_frame().T], ignore_index=True)

# Gegenseitige Wunschgruppen finden (nur 2er-Gruppen)
mutual_groups = set()
for spieler, wünsche in wunsch_dict.items():
    for gewünschter in wünsche:
        if gewünschter in wunsch_dict and spieler in wunsch_dict[gewünschter]:
            gruppe = tuple(sorted([spieler, gewünschter]))
            mutual_groups.add(gruppe)

# Umwandlung in Listen
verified_groups = [list(gruppe) for gruppe in mutual_groups]
```

```python
# einseitige Erwähnung reicht
group_to_merge = verified_groups.copy()

def merge_groups(groups):
    merged = []

    for group in groups:
        added = False
        for mgroup in merged:
            if any(elem in mgroup for elem in group):
                mgroup.update(group)
                added = True
                break
        if not added:
            merged.append(set(group))

    return [list(mgroup) for mgroup in merged]

merged_groups = merge_groups(group_to_merge)

merged_groups
```

### Anzahl und Rostergröße der Ligen festlegen

```python
import random
import pandas as pd
import datetime

# === VORBEREITUNG ===

# Mapping 'index' zu 'Sleeper' anzeigenamen
sleeper_to_show = df_slrleagues.set_index('index')['Sleeper'].to_dict()

# Liste aller Commishs und Nicht-Commishs
commish_owners = df_slrleagues[df_slrleagues['Commish'] == True]['index'].tolist()
non_commish_owners = df_slrleagues[df_slrleagues['Commish'] == False]['index'].tolist()

# Validierung
if len(commish_owners) < len(leagues):
    raise ValueError("Nicht genügend Commishs für jede Liga vorhanden.")

# Hilfsfunktion: Gibt Mitglieder einer Gruppe zurück
def get_group_members(owner):
    for group in merged_groups:
        if owner in group:
            return random.sample(group, k=len(group))  # zufällige Reihenfolge
    return [owner]  # falls keine Gruppe vorhanden

# === LIGEN INITIALISIEREN ===

leagues_dict = {}
assigned_owners = set()
commish_assigned = set()

# Commish zufällig mischen
random.shuffle(commish_owners)

# === COMMISHS VERTEILEN (inkl. Gruppen) ===

for league in leagues:
    for commish in commish_owners:
        if commish in assigned_owners:
            continue

        group = get_group_members(commish)
        unassigned_group = [m for m in group if m not in assigned_owners]

        if len(unassigned_group) <= league_size[leagues.index(league)]:
            leagues_dict[league] = [sleeper_to_show[m] for m in unassigned_group]
            assigned_owners.update(unassigned_group)
            commish_assigned.add(commish)
            break

# === RESTLICHE SPIELER (inkl. überschüssige Commishs) VORBEREITEN ===

# Alle nicht zugewiesenen Owner (Commish + normale Spieler)
unassigned_owners = [
    owner for owner in (commish_owners + non_commish_owners)
    if owner not in assigned_owners
]
random.shuffle(unassigned_owners)

# === RESTLICHE SPIELER EINTEILEN UNTER BERÜCKSICHTIGUNG IHRER GRUPPEN ===

for league in leagues:
    league_owners = leagues_dict.get(league, [])

    while len(league_owners) < league_size[leagues.index(league)] and unassigned_owners:
        current_owner = unassigned_owners.pop(0)

        if current_owner in assigned_owners:
            continue

        group = get_group_members(current_owner)
        unassigned_group = [m for m in group if m not in assigned_owners]

        if len(league_owners) + len(unassigned_group) <= league_size[leagues.index(league)]:
            league_owners.extend(sleeper_to_show[m] for m in unassigned_group)
            assigned_owners.update(unassigned_group)
        else:
            # Gruppe passt nicht – später erneut versuchen
            unassigned_owners.append(current_owner)

        leagues_dict[league] = league_owners

# === LEERE PLÄTZE OHNE GRUPPENZUGEHÖRIGKEIT AUFFÜLLEN ===

remaining_owners = [
    owner for owner in (commish_owners + non_commish_owners)
    if owner not in assigned_owners
]

for league, size in zip(leagues, league_size):
    while len(leagues_dict[league]) < size and remaining_owners:
        leagues_dict[league].append(sleeper_to_show[remaining_owners.pop(0)])

for league in leagues_dict:
    random.shuffle(leagues_dict[league])

# === LIGEN AUF GLEICHE LÄNGE BRINGEN ===

max_league_size = max(len(v) for v in leagues_dict.values())
for league in leagues_dict:
    while len(leagues_dict[league]) < max_league_size:
        leagues_dict[league].append('')

# === DATAFRAME ERZEUGEN ===

df_ligen = pd.DataFrame(leagues_dict)
df_ligen.index += 1
supabase_df = df_ligen.T.reset_index(drop=True)
supabase_df.index += 1
supabase_df['league_name'] = "SLR2025 - Liga " + supabase_df.index.astype(str)
df_ligen.index.name = 'Draftposition'

# === EXPORTIEREN ===

current_time = datetime.datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
filename = f'SLR2025-Auslosung_{current_time}'

df_ligen.to_csv(f'{filename}.csv', index=False)
# df_ligen.to_excel(f'{filename}.xlsx', index=False)  # Optional

supabase_df.index.name = "league"
supabase_df = supabase_df.reset_index(level=['league'])
```
